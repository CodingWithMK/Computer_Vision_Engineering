# Week 2 - Lesson 1 - Histogram & Thresholding

````python
import cv2
import matplotlib.pyplot as plt
import numpy as np
````

### 1. ```cv2.calcHist()```

- Calculates the histogram of an image (or of the mask of an area).
- **Signature**: ```cv2.calcHist(images, channels, mask, histSize, ranges)```

````python
img = cv2.imread("test_image.jpg", cv2.IMREAD_GRAYSCALE)
hist = cv2.calcHist([img], [0], None, [256], [0, 256])

plt.plot(hist)
plt.title("Histogram of Grayscale Image")
plt.show()
````

**NOTE**: Here we can see that our test image of a road has the most pixel values (over 20k pixels) in a range of the values 75-90 (dark gray)

### 2. ```cv2.equalizeHist()```

- Normalizes the histogram of an one channel (grayscale) image by increasing its contrast
- **Signature**: ```cv2.equalize(src)```

````python
eq = cv2.equalizeHist(img)
plt.imshow(eq, cmap='gray')
plt.title("Equalized Grayscale Image")
plt.show()
````

### 3. ```cv2.filter2D()```

- Performs convolution on an image with a customized kernel.
- **Signature**: ```cv2.filter2D(src, ddepth, kernel)```

````python
kernel = np.ones((5, 5), np.float32) / 25
filtered_img = cv2.filter2D(img, -1, kernel)

plt.imshow(filtered_img, cmap='gray')
plt.title("Filtered Image (Filter2D - Mean Blur)")
plt.show()
````

### 4. ```cv2.GaussianBlur()```

- Applies a Gaussian blur to reduce noise in the image (using Gauss Distribution)
- **Signature**: ```cv2.GaussianBlur(src, ksize, sigmaX)```

````python
gaussian_blur = cv2.GaussianBlur(img, (5, 5), 0)

plt.imshow(gaussian_blur, cmap='gray')
plt.title("Gaussian Blurred Image")
plt.show()
````

### 5. ```cv2.medianBlur()```

- Applies median filter to the image, effective against salt-and-pepper noise
- **Signature**: ```cv2.medianBlur(src, ksize)```

````python
median_blur = cv2.medianBlur(img, 5)

plt.imshow(median_blur, cmap='gray')
plt.title("Median Blurred Image")
plt.show()
````

### 6. ```cv2.blur()```

- Applies mean blur to the image.
- **Signature**: ```cv2.blur(src, ksize)```

````python
median_blur = cv2.medianBlur(img, 5)

plt.imshow(median_blur, cmap='gray')
plt.title("Median Blurred Image")
plt.show()
````
